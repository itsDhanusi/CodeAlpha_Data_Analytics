```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("books_data.csv")

# Convert price into numeric values
df["Price"] = (
    df["Price"].astype(str)
    .str.replace("£", "", regex=False)
    .str.replace("Â", "", regex=False)
    .astype(float)
)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Price"], bins=10, edgecolor="black")
plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Rating")
plt.title("Number of Books by Rating")
plt.xlabel("Book Rating")
plt.ylabel("Number of Books")
plt.show()
```

```python
top_books = df.nlargest(10, "Price")

plt.figure(figsize=(10, 6))
plt.barh(top_books["Book Name"], top_books["Price"])
plt.title("Top 10 Highest-Priced Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Name")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Price"], bins=10, edgecolor="black")
plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.savefig("book_price_chart.png", bbox_inches="tight")
plt.show()

files.download("book_price_chart.png")
```
